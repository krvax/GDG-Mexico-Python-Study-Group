## **CÁLCULO DE UNA VARIABLE**

El **cálculo infinitesimal** (también llamado unicamente cálculo) es una rama de las matemáticas que se centra en el estudio del cambio continuo a través del concepto de infinitésimos.

Es tópico tiene dos ramas principales:

*   **Cálculo diferencial**: se enfoca en las tasas de cambio y las pendientes de las curvas.

*   **Cálculo integral**: se ocupa de la acumulación de cantidades y las áreas bajo y entre curvas.

El cálculo es una herramienta fundamental en muchas áreas de la ciencia y la ingeniería, ya que **permite modelar y analizar fenómenos que cambian continuamente**.

Sabiendo que:

\begin{equation}
y = f(x)
\, \, \, \, \, \, \, \, \, \, \, \,
\Rightarrow
\, \, \, \, \, \, \, \, \, \, \, \,
y' = f'(x) = \frac{\mathrm{d} y}{\mathrm{d} x}
\, \, \, \, \, \, \, \, \, \, \, \,
∧
\, \, \, \, \, \, \, \, \, \, \, \,
F(x) = \int  f(x)dx + C
\end{equation}

````python
# Importar librerías

import numpy as np
import matplotlib.pyplot as plt
````

````python
# Definir variables globales

x = np.linspace(-10, 10, 1000)
len(x)
````

````python
# Definir función de derivación

def derivada(y):
  dy = np.diff(y)/np.diff(x)
  return dy

def derivada(y):
  return np.diff(y)/np.diff(x)

derivada = lambda y: np.diff(y)/np.diff(x)
````

````python
len(derivada(x**2))
````

````python
# Definir función de antiderivación

def antiderivada(y):
  sy = np.cumsum((y[1:] + y[:-1]) * np.diff(x)/2)
  return sy

def antiderivada(y):
  return np.cumsum((y[1:] + y[:-1]) * np.diff(x)/2)

antiderivada = lambda y: np.cumsum((y[1:] + y[:-1]) * np.diff(x)/2)
````

````python
len(antiderivada(x**2))
````

````python
# Definir función de visualización

def visualizacion(y, dy, sy):
  plt.plot(x, y, color = 'blue', label = 'Función')
  plt.plot(x[1:], dy, color = 'green', label = 'Derivada')
  plt.plot(x[1:], sy, color = 'purple', label = 'Antiderivada')
  plt.axhline(y = 0, color = 'red')
  plt.axvline(x = 0, color = 'red')
  plt.legend(loc = 'upper left')
  plt.grid()
  plt.show()
````
