```python
import pandas as pd
import matplotlib.pyplot as plt

# Configuración para ver gráficos directamente en el notebook
%matplotlib inline
```

## 1. Cargar y Explorar el Dataset

El primer paso en ciencia de datos es cargar los datos. Usaremos `pd.read_csv()` para leer el archivo `stats.csv` que contiene el rendimiento de los jugadores en la última sesión.

```python
# Cargar los datos del archivo CSV local
df = pd.read_csv('stats.csv')

# Mostrar las primeras filas del dataset
df.head()
```

## 2. Cálculo de Métricas Clave: Efectividad de Tiro

La efectividad de tiro de un jugador es el porcentaje de tiros encestados respecto a los intentados:

$$\text{Efectividad (\%)} = \frac{\text{shots\_made}}{\text{shots\_attempted}} \times 100$$

En Python, realizar operaciones vectorizadas entre columnas con Pandas es tan sencillo como operar variables básicas.

```python
# Calcular el porcentaje de efectividad de tiro
df['shooting_efficiency'] = (df['shots_made'] / df['shots_attempted']) * 100

# Redondear a un decimal para mejor lectura
df['shooting_efficiency'] = df['shooting_efficiency'].round(1)

# Mostrar el DataFrame actualizado
df
```

## 3. Estadísticas Descriptivas del Equipo

¿Cuál es el promedio general de rebotes y asistencias del equipo? Podemos usar las funciones matemáticas integradas de Pandas como `.mean()` y `.max()`.

```python
avg_rebounds = df['rebounds'].mean()
avg_assists = df['assists'].mean()
max_efficiency = df['shooting_efficiency'].max()

print(f"Promedio de rebotes por jugador: {avg_rebounds:.2f}")
print(f"Promedio de asistencias por jugador: {avg_assists:.2f}")
print(f"Efectividad de tiro máxima registrada: {max_efficiency}%")
```

## 4. Filtrar y Ordenar Datos

Queremos ver quiénes son los jugadores más eficientes (efectividad mayor al 60%) y ordenarlos de forma descendente.

```python
# Filtrar jugadores con efectividad mayor al 60%
efficient_players = df[df['shooting_efficiency'] > 60.0]

# Ordenar de forma descendente por efectividad
efficient_players_sorted = efficient_players.sort_values(by='shooting_efficiency', ascending=False)

efficient_players_sorted
```

## 5. Visualizar los Datos

Un gráfico vale más que mil tablas. Grafiquemos la efectividad de tiro de cada jugador usando un gráfico de barras de Matplotlib.

```python
# Configurar el tamaño del gráfico
plt.figure(figsize=(10, 6))

# Crear el gráfico de barras
plt.bar(df['player'], df['shooting_efficiency'], color='coral', edgecolor='black')

# Personalizar etiquetas y título
plt.title('Efectividad de Tiro por Jugador - Cóatl Baloncesto', fontsize=14, fontweight='bold')
plt.xlabel('Jugador', fontsize=12)
plt.ylabel('Efectividad (%)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Mostrar el gráfico
plt.show()
```
